Три дополнительных серии к LatinISE: без Вульгаты, без Codex и Institutiones Юстиниана, без обоих блоков. Исходные модели не меняются. По умолчанию 30 обучений mc5; mc50 можно выполнить отдельно. Используется существующий `src.core`. Исходные тексты должны быть локально доступны в `data/new_lemmatized_texts/` (при необходимости `python download_corpus.py`).

```python
from pathlib import Path
import sys

PROJECT_ROOT = None  # например r"D:\Research\latin-embeddings-models"
MIN_COUNTS = (5,)  # (5, 50) для обеих серий
PERIODS = None  # все три среза; например ["post180_no_vulgate"]
INCLUDE_EARLY_AUDIT = True

candidates = [Path(PROJECT_ROOT)] if PROJECT_ROOT else [Path.cwd(), Path.cwd().parent, Path.cwd()/"latin-embeddings-models"]
ROOT = next((p.resolve() for p in candidates if (p/"src/ablation.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Укажите PROJECT_ROOT")
sys.path.insert(0, str(ROOT))
from src import core, ablation
from IPython.display import display
print(ROOT)
```

## 1. Состав контрольных срезов и частотность

Проверяем состав исходных срезов по метаданным и полные частотные словари по ранее опубликованной статистике. Все объёмы считаются заново функцией `core.tokens`. ## 2. Обучение

Параметры берутся из `config/settings.json`, срезы не перемешиваются с исходными.
